```python
import mne
```

```python
subject = 'sub-01'
session = 'ses-01'
```

```python
# 파일 경로
file_path = f'D:/Users/windows/Desktop/Inner Speech Decoding/Dataset/{subject}_{session}_eeg-epo.fif'
```

```python
# 데이터 불러오기
epochs = mne.read_epochs(file_path)
```

```python
# 데이터 확인 (epoch 정보 확인)
print(epochs.info)
```

```python
# X 데이터 (EEG 신호)와 y 데이터 (레이블) 분리
X = epochs.get_data()  # EEG 신호
y = epochs.events[:, -1]  # 레이블 (이벤트 정보)
```

```python
from tensorflow.keras.models import load_model
import numpy as np
```

```python
# EEGNet 모델 불러오기 (모델이 이미 학습되어 있어야 함)
model = load_model("D:/Users/windows/Desktop/arl-eegmodels-master/examples/EEGNet-8-2-weights.h5")
```

```python
# 모델 예측
y_pred = model.predict(X)
```

```python
from sklearn.metrics import classification_report

# 모델 성능 평가
print(classification_report(y, y_pred.argmax(axis=1)))  # 다중 클래스일 경우
```
