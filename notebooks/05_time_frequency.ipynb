Run after installing `pip install -e ".[dev]"`. Configure data paths first.
Historical notebooks remain untouched in the two original folders.

```python
from pathlib import Path
import os
from inner_speech.config import load_config
from inner_speech.cli import run_stage
# Start Jupyter in the repository or notebooks directory.
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").exists())
config_path = Path(os.environ.get("INNER_SPEECH_CONFIG", root / "configs/default.yaml"))
cfg = load_config(config_path)
```

```python
import mne
from inner_speech.time_frequency import multitaper_power
folder = Path(cfg['output_root']) / cfg['subjects'][0] / cfg['sessions'][0]
task = mne.read_epochs(folder / 'task-epo.fif', preload=True)
rest = mne.read_epochs(folder / 'rest-epo.fif', preload=True)
task_power, rest_power = multitaper_power(task), multitaper_power(rest)
# Windows have different event anchors. This is descriptive, not a prestimulus baseline.
(task_power - rest_power).plot(picks=[0], title='Task minus inter-trial rest power')
```
